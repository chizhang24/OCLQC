```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
naive_df = pd.read_csv('naive.csv')
replay_df = pd.read_csv('replay.csv')

avg_naive = naive_df.iloc[:, 2].mean()
avg_replay = replay_df.iloc[:, 2].mean()

print(f'Naive Average Accuracy: {avg_naive}')
print(f'Replay Average Accuracy: {avg_replay}')
```

```python
x_naive, y_naive = naive_df.iloc[:, 1], naive_df.iloc[:, 2]
x_replay, y_replay = replay_df.iloc[:, 1], replay_df.iloc[:, 2]

# 3. Create the Plot
plt.figure(figsize=(12, 6))

# Plot raw accuracy with low alpha to show the "noise"
plt.plot(x_naive, y_naive, color='black', alpha=0.3, label='Naive (avg acc: {:.4f})'.format(avg_naive))
plt.plot(x_replay, y_replay, color='blue', alpha=0.3, label='Replay (avg acc: {:.4f})'.format(avg_replay))

# # Plot a Moving Average (window of 10) to see the drift trend clearly
# plt.plot(x_naive, y_naive.rolling(window=10).mean(), 
#          color='blue', linewidth=2, label='Naive (Trend)')
# plt.plot(x_replay, y_replay.rolling(window=10).mean(), 
#          color='red', linewidth=2, label='Replay (Trend)')

# 4. Formatting
plt.title('Qubit 0/1 Classification: Naive vs. Replay (7 Days)', fontsize=14)
plt.xlabel('Time Step (30-minute windows)', fontsize=12)
plt.ylabel('Test Accuracy', fontsize=12)
plt.ylim(0.8, 0.95) # Accuracy is between 0 and 1
plt.grid(True, which='both', linestyle='--', alpha=0.5)
plt.legend(loc='lower right')



plt.tight_layout()
plt.savefig('qubit_accuracy_comparison.png', dpi=300)
plt.show()
```
